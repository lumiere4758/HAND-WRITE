```python
import numpy, pickle
import matplotlib.pyplot as plt
```

```python
with open('dataset/data.pickle', 'rb') as f:
    data = pickle.load(f)
```

```python
m = 10     # rows
n = 14     # columns
o = 210  # offset
fig, ax = plt.subplots(m, n, sharex=False, sharey=False, figsize=(n*2,m*2))
for i in range(m):
    for j in range(n):
        ax[i,j].imshow(data['img'][i*n + j + o])
        ax[i,j].set_title(data['label'][i*n + j + o])
        ax[i,j].set_xticks([])
        ax[i,j].set_yticks([])
```
